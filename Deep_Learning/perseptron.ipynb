```python
import matplotlib.pyplot as plt
import random
import numpy as np

# User Inputs
minX = int(input("Enter minimum X value: ")) 
maxX = int(input("Enter maximum X value: "))
n = int(input("Number of points per group: "))

x_top, y_top = [], []
x_bottom, y_bottom = [], []

slope = 1.5 
gap = 100 
noise = 25

# Generating Synthetic Data
for _ in range(n):
    # Top group (Target -1)
    rx1 = random.randint(minX, maxX)
    ry1 = (slope * rx1 + gap) + random.randint(-noise, noise)
    x_top.append(rx1)
    y_top.append(ry1)

    # Bottom group 
    rx2 = random.randint(minX, maxX)
    ry2 = (slope * rx2) + random.randint(-noise, noise)
    x_bottom.append(rx2)
    y_bottom.append(ry2)

# Preparing data for training
data = []
for i in range(len(x_top)):
    data.append([x_top[i], y_top[i], -1]) 
for j in range(len(x_bottom)):
    data.append([x_bottom[j], y_bottom[j], 1])

# Perceptron Training Part
iteration = int(input("Enter number of iterations: "))
w1 = random.uniform(-1, 1) 
w2 = random.uniform(-1, 1)
b = random.uniform(-1, 1)
learning_rate = 0.0001 

for t in range(iteration):
    correct_count = 0 
    for k in range(len(data)):
        x_val = data[k][0]
        y_val = data[k][1]
        target = data[k][2]

        # result = w1*x + w2*y + b
        result = x_val * w1 + y_val * w2 + b
        
        predicted = 1 if result > 0 else -1
        
        if target == predicted:
            correct_count += 1
        else:
            # Update weights if prediction is wrong
            error = target - predicted
            w1 = w1 + learning_rate * error * x_val
            w2 = w2 + learning_rate * error * y_val
            b = b + learning_rate * error

print(f"Correct predictions: {correct_count} (Total: {len(data)})")

# Visualisation
plt.scatter(x_top, y_top, color='blue', label='Top Group (-1)', alpha=0.7)
plt.scatter(x_bottom, y_bottom, color='green', label='Bottom Group (1)', alpha=0.7)

# Drawing the Red Decision Boundary
# The boundary is where w1*x + w2*y + b = 0
x_boundary = np.array([minX, maxX])
if w2 != 0:
    y_boundary = (-w1 / w2) * x_boundary - (b / w2)
    plt.plot(x_boundary, y_boundary, color='red', linewidth=2, label='Decision Boundary')

plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.title("Perceptron Linear Classification")
plt.legend()
plt.grid(True, linestyle='--')
plt.savefig("Perceptron_Result.png")
plt.show()
```
