```python
import matplotlib.pyplot as plt
import random
import math

# User input for number of clusters
num_clusters = int(input('How many clusters K do you want to create?: '))

# Generate 100 random data points
x_points = []
y_points = []
for i in range(100):
    x_points.append(random.randint(1, 100))
    y_points.append(random.randint(1, 100))

# Initialize random centroids
centroid_x = []
centroid_y = []
for i in range(num_clusters):
    centroid_x.append(random.randint(0, 100))
    centroid_y.append(random.randint(0, 100))

# Define colors for clusters
colors = ['blue', 'green', 'yellow', 'purple', 'orange', 'cyan', 'magenta']

while True:
    assigned_clusters = []

    for i in range(len(x_points)):
        distances = []
        for j in range(len(centroid_x)):
            # Euclidean distance calculation
            d = math.sqrt((x_points[i] - centroid_x[j])**2 + (y_points[i] - centroid_y[j])**2)
            distances.append(d)
        
        nearest_centroid_index = distances.index(min(distances))
        assigned_clusters.append(nearest_centroid_index)

    new_centroid_x = []
    new_centroid_y = []

    # Calculate the mean of points in each cluster
    for k in range(num_clusters):
        cluster_x_points = [x_points[i] for i in range(len(x_points)) if assigned_clusters[i] == k]
        cluster_y_points = [y_points[i] for i in range(len(y_points)) if assigned_clusters[i] == k]
        
        if len(cluster_x_points) > 0:
            new_centroid_x.append(sum(cluster_x_points) / len(cluster_x_points))
            new_centroid_y.append(sum(cluster_y_points) / len(cluster_y_points))
        else:
            # If a cluster is empty, keep the previous centroid
            new_centroid_x.append(centroid_x[k])
            new_centroid_y.append(centroid_y[k])

    # Check for convergence
    if new_centroid_x == centroid_x and new_centroid_y == centroid_y:
        print("Convergence reached. Plotting results")
        break
        
    centroid_x = new_centroid_x
    centroid_y = new_centroid_y

# Visualization
plt.figure(figsize=(10, 6))

# Plot data points colored by cluster
for i in range(len(x_points)):
    plt.scatter(x_points[i], y_points[i], color=colors[assigned_clusters[i] % len(colors)], alpha=0.5)

# Plot final centroids as black stars
plt.scatter(centroid_x, centroid_y, color='black', marker='*', s=250, label='Centroids')

plt.title(f"K-Means Clustering Result K={num_clusters}")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import random
import math

# クラスタ数の入力
k_suu = int(input('クラスタ数 K を入力してください: '))

# 100個のランダムなデータポイントを作成
x_ten = []
y_ten = []
for i in range(100):
    x_ten.append(random.randint(1, 100))
    y_ten.append(random.randint(1, 100))

# 初期中心点を ランダムに設定
chushin_x = []
chushin_y = []
for i in range(k_suu):
    chushin_x.append(random.randint(0, 100))
    chushin_y.append(random.randint(0, 100))

# クラスタ用の色設定
iro_list = ['blue', 'green', 'yellow', 'purple', 'orange', 'cyan', 'magenta']

while True:
    wariate_list = []
    
    for i in range(len(x_ten)):
        kyori_list = []
        for j in range(len(chushin_x)):
            # ユークリッド距離の計算
            d = math.sqrt((x_ten[i] - chushin_x[j])**2 + (y_ten[i] - chushin_y[j])**2)
            kyori_list.append(d)
        
        saiteki_chushin = kyori_list.index(min(kyori_list))
        wariate_list.append(saiteki_chushin)

    atarashii_chushin_x = []
    atarashii_chushin_y = []

    # 更新フェーズ：各クラスタの平均値を計算して中心点を移動
    for k in range(k_suu):
        kurasta_x = [x_ten[i] for i in range(len(x_ten)) if wariate_list[i] == k]
        kurasta_y = [y_ten[i] for i in range(len(y_ten)) if wariate_list[i] == k]
        
        if len(kurasta_x) > 0:
            atarashii_chushin_x.append(sum(kurasta_x) / len(kurasta_x))
            atarashii_chushin_y.append(sum(kurasta_y) / len(kurasta_y))
        else:
            atarashii_chushin_x.append(chushin_x[k])
            atarashii_chushin_y.append(chushin_y[k])

    # 中心点が動かなくなったら終了
    if atarashii_chushin_x == chushin_x and atarashii_chushin_y == chushin_y:
        print("収束しました。結果を表示します。")
        break
        
    chushin_x = atarashii_chushin_x
    chushin_y = atarashii_chushin_y

# 可視化
plt.figure(figsize=(10, 6))

# クラスタごとに色分けしてプロット
for i in range(len(x_ten)):
    plt.scatter(x_ten[i], y_ten[i], color=iro_list[wariate_list[i] % len(iro_list)], alpha=0.5)

# 最終的な中心点を黒い星でプロット
plt.scatter(chushin_x, chushin_y, color='black', marker='*', s=250, label='中心点（セントロイド）')

plt.title(f"K-Means クラスタリング結果 K={k_suu}")
plt.xlabel("X 軸")
plt.ylabel("Y 軸")
plt.legend()
plt.grid(True)
plt.show()
```
